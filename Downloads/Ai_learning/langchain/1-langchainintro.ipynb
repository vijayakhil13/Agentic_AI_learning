```python
import os
from dotenv import load_dotenv
load_dotenv()
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
from langchain.agents import create_agent
from langchain_groq import ChatGroq
llm= ChatGroq(
    model="openai/gpt-oss-120b",
    groq_api_key= os.getenv("GROQ_API_KEY")
)
def get_weather(city:str)-> str:
    """get weather of hyderabad"""
    return "32 degrees sunny"
agent=create_agent(
    model=llm,
    tools=[get_weather],
    system_prompt="You are a helpful assistant that uses tools to answer weather-related questions",
)
agent
respose=agent.invoke({"messages":[{"role":"user", "content": "what is weather in hyderabad"}]})
respose["messages"][-1].content
```
